```python
tiktok_account = 'https://www.tiktok.com/@kelseysoles'
```

```python
driver=webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get(tiktok_account)
```

```python
download_vids =[]
elements = driver.find_elements(By.TAG_NAME, 'a')
for i,element in enumerate(elements):
    if '/video/' in element.get_attribute('href'):
        print(f'{i}:{element.get_attribute('href')}')
        if element.get_attribute('href') not in download_vids:
            download_vids.append(element.get_attribute('href'))
```

```python
#download_the video 
vid_save =[]
for link in download_vids:
    linka = link.split('video/')
    vid_save.append(linka[1])    
```

```python



def get_tiktok_json(video_url,browser_name=None):
    
    global cookies
    if browser_name is not None:
        cookies = getattr(browser_cookie3,browser_name)(domain_name='www.tiktok.com')
    tt = requests.get(video_url,
                      headers=headers,
                      cookies=cookies,
                      timeout=20)
    # retain any new cookies that got set in this request
    cookies = tt.cookies
    soup = BeautifulSoup(tt.text, "html.parser")
    tt_script = soup.find('script', attrs={'id':"SIGI_STATE"})
    try:
        tt_json = json.loads(tt_script.string)
    except AttributeError:
        return
    return tt_json 
```
